## Datasetin yüklənməsi və kəşf edilməsi

Dataseti yükləyin və ilk 20 dəyərini göstərin

```python
df=pd.read_csv(r'/Users/xayalisgenderov/Downloads/Fifa 2019.csv')
df.head(20)
```

Baxin burda Birinci sutunda Unnamed deye bir sey var onu yığışdıraq

```python
df=df.drop('Unnamed: 0',axis=1)
```

## Preprocessing (Oyunçu dəyərinin rəqəmə çevrilməsi)

##### Value sutununa baxmaq isteyirem. Sutunun ilk 10 reqemini nece gore bilerem?

```python
df[['Value']].head(10)
```

##### Burda Euro isaresini cixaririq, M herfini 1000000-la, K herfini ise 1000-le evezleyirik

```python
df['Value']=df['Value'].str.replace('€', '', regex=False)
df['Value']=df['Value'].str.replace('M', '*1000000', regex=False)
df['Value']=df['Value'].str.replace('K', '*1000', regex=False)
df['Value'].head(10)
```

##### İndi isə sütunu iki hissəyə böləcək kod yazın və bu iki sütunu bir birinə vuraraq real "Value" sütununu əldə edin (float tipində)

```python
value1=df['Value'].str.split('*', expand=True)
value1[0]=value1[0].astype(float)
value1[1]=value1[1].fillna(1).astype(float)
df['Value']=value1[0]*value1[1]
df['Value'].head(10)
```

##### Menim sutunumda null deyerler ola biler. Men isteyirem ki bunu 0 ile evez edim

```python
df['Value']=df['Value'].fillna(0)
df['Value'].isnull().sum()
```

## Visualization and Analysis

#####  Oyunçuların Pozisiya paylanması (Categorical)

Bir fiqur (10,10) yaradın Bu fiqur daxilində hansı pozisiyadan neçə nəfər uyğun chart vasitəsilə göstərin ("Position" sütunu)

```python
plt.figure(figsize=(10,10))
sns.countplot(x='Position',data=df,order=df['Position'].value_counts().index)
plt.xticks(rotation=90)
plt.title('Oyunçuların pozisiyaları')
plt.show()
```

#### Oyunçuların yaş paylanması (Numeric)

Bir fiqur (10,10) yaradın Bu fiqur daxilində hansı yaş sütununu uyğun chart vasitəsilə göstərin ("Age" sütunu)

```python
plt.figure(figsize=(10,10))
sns.histplot(df['Age'],bins=10)
plt.title('Oyunçuların yaş paylanması')
plt.show()
```

Bir də bunu kde plot vasitəsilə verin

```python
plt.figure(figsize=(10,10))
sns.kdeplot(df['Age'],fill=True)
plt.title('Yaş paylanması - KDE plot')
plt.show()
```

###  Oyunçuların istifadə etdiyi ayağının qiymətə təsiri

##### Birinci oyuncularin hansi ayagini istifade etdiyini oyrenek. Yeni hansi ayagindan (sol ve ya sag) istifade eden nece neferdi?

```python
df['Preferred Foot'].value_counts()
```

##### Bir de bunu qrafik kimi gorek. Kateqoriya ucun hansi plotdan istifade etmeliyem?

```python
plt.figure(figsize=(10,10))
sns.countplot(x='Preferred Foot',data=df)
plt.title('İstifade olunan ayağ')
plt.show()
```

##### Bunu bir də pie chart kimi görmək istəyirəm ki, paylar necə bölünür?

```python
plt.figure(figsize=(7,7))
plt.pie(df['Preferred Foot'].value_counts(),labels=df['Preferred Foot'].value_counts().index,autopct='%1.1f%%')
plt.title('Ayağa görə paylanma')
plt.show()
```

##### Qruplaşdırma funksiyalarından istifadə edərək istifadə edilən ayağa görə ortalama qiyməti tapın

```python
df.groupby('Preferred Foot')['Value'].mean()
```

## Dashboard

Bir fiqur yaradın (15,15). Daha sonra 4 ədəd subplot yaradın və yuxarıda qurduğunuz 4 plotu bu dashboard-a yerləşdirin
